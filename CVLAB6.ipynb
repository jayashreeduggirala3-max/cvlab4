```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show_comparision(original, sliced_no_bg, sliced_with_bg):
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(original, cmap='gray')
    plt.title('Original Image')
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.imshow(sliced_no_bg, cmap='gray')
    plt.title('Sliced (No Background)')
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(sliced_with_bg, cmap='gray')
    plt.title('Sliced (With Background)')
    plt.axis('off')

    plt.show()

img = cv2.imread("/content/gls.jpg", cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Error: Could not open or find the image. Check your image path!")
else:

    rmin = 100
    rmax = 200
    sliced_no_bg = np.zeros(img.shape)

    sliced_with_bg = img.copy()
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
          pixel_value = img[i] [j]
          if rmin <= pixel_value<= rmax:
                sliced_no_bg[i][j] = 255
                sliced_with_bg[i][j] = 255
          else:
                sliced_no_bg[i][j] = 0

    show_comparision(img,sliced_no_bg,sliced_with_bg)
```
